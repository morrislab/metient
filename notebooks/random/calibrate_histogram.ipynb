```python
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np

OUTPUT_DIR = '/lila/data/morrisq/divyak/projects/metient/metient/jupyter_notebooks/random/outputs'

fig, ax = plt.subplots(1, 1, figsize=(1.5, 1))
# sns.set(style='ticks', font="Arial", rc={'axes.labelsize': 14, 'xtick.labelsize': 12, 'ytick.labelsize': 12, 'axes.linewidth': 1.2})

# Create bimodal distribution
x = np.concatenate([np.random.normal(1.5, 0.4, 500), np.random.normal(2.5, 0.4, 500)])
x = np.random.normal(1.7, 0.4, 1000)

# Create unimodal distribution 
y = np.random.normal(2, 0.3, 1000)
y = y.tolist()

from scipy.stats import gaussian_kde
kde_x = gaussian_kde(x)
kde_y = gaussian_kde(y)
x_range = np.linspace(min(min(x), min(y)), max(max(x), max(y)), 200)
# Plot the first distribution slightly higher
ax.fill_between(x_range, kde_y(x_range) + 0.2, 0.2, alpha=0.8, color='#893071')

# Plot the second distribution at base level
ax.fill_between(x_range, kde_x(x_range), 0, alpha=0.8, color='#E67B61')

for spine in ax.spines:
    ax.spines[spine].set_visible(False)

# plt.legend(['Metastasis prior', 'Parsimony'], frameon=False, loc='center left', bbox_to_anchor=(1, 0.5), title='Distribution',fontsize=12,title_fontsize=12)
ax.set_xticks([])
ax.set_yticks([])
ax.set_xlabel('')
ax.set_ylabel('')

plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_DIR, 'calibrate_histogram.svg'), bbox_inches='tight', dpi=300)
plt.show()
```
